```python
# Alışveriş sepeti
urun1 = float(input("1. ürününüzün fiyatını giriniz :"))
urun2 = float(input("2. ürününüzün fiyatını giriniz :"))
urun3 = float(input("3. ürününüzün fiyatını giriniz :"))

top_fiy = float( urun1 + urun2 + urun3)

if top_fiy > 200:
    print("Ürünlerinize %10 indirim uygulanmıştır.")
    top_fiy= top_fiy - (top_fiy * 0.1)

print("Toplam fiyatınız : ", top_fiy)
    
```

```python
# Harf frekansı (string)
kelime = input("Bir kelime giriniz : ")

sonuc = {}
for harf in kelime :
    sonuc[harf] = sonuc.get(harf,0) + 1 #get fonk sözlükten harflerialır 
print(sonuc)
```

```python
# Şifre kontrolü

sifre = input("Bir şifre giriniz :")

if len(sifre) < 8 : 
    print("Şifrenin 8 karakterden uzun olmalı!")
elif not any(harf.isupper() for harf in sifre):
    print("Şifreniz en az bir büyük harf içermeli !")
elif not any(harf.isdigit() for harf in sifre):
    print("Şifreniz en az bir rakam içermeli !")
else:
    print("Şifreniz geçerlidir.")

```

```python
liste = [12,4,9,25,30,7,18]

liste_np = np.array(liste)
ort = liste_np.mean()
print(f'listenin ortalaması {ort} dır')

ort_buyuk = []
for i in liste:
    if i>ort:
        ort_buyuk.append(i)
print(f'Ortalamadan büyük sayılar :{ort_buyuk}')


```

```python
# nested loop (DESEN)

for i in range (1,6): # satır sayısı
    for j in range(i): # her satırda i kadar yıldız 
        print("*",end="")# alt satıra geçmeden yıldızları yan yana yazdırmayı sağlar
    print() #tekrar bir alt satıra geç
        
```

```python
# while döngüsü
toplam = 0
sayac = 0
while True: # sonsuz döngü başlat
  sayi = int(input("Lütfen bir sayı giriniz : "))
  toplam += sayi 
  sayac += 1
    
  if sayi == 0: # sayı sıfıra eşit olunca çık
      print("0 girdiniz program durdurulacak")
      break


if sayac > 0 : # girilen sayı kadar işlem yap
    ort = toplam / sayac
    print("Girilen sayıların toplamı:", toplam)
    print("Girilen sayı adedi:", sayac)
    print("Sayıların ortalaması:", ort)
else:
    print("Hiç sayı girilmedi.")


```

```python
# palindrom kntrolü 
kelime = input("Lütfen bir kelime giriniz :")

if kelime == kelime[::-1]: # kelimenin tersten yazılışı ve kendisi eşit öi kontrolü
    print("Girdiğiniz kelime palindromdur.",kelime)
else:
    print("Girdiğiniz kelime palindrom değildir.",kelime)
    

    
```

```python
# List Comprehension
liste=[]

for i in range (1,101):
    if i % 3 == 0 :
        if i % 5 == 0:
            sonuc = i*i
            liste.append(sonuc)

print("1'den 100'e kadar olan sayıların, \n hem 3'e hem 5'e bölünebilenlerin karesi : ",liste)

            
```

```python
cumle_sayisi = int(input("Kaç yorum cümlesi gireceksiniz ? "))
liste = []


for i in range(cumle_sayisi):
    cumle = input(f"{i+1}. cümleyi giriniz: ") 
    liste.append(cumle)

kelimeler = cumle.split(" ")
kac_iyi = kelimeler.count("iyi")

en_uzun = max(liste, key=len)
en_kisa = min(liste, key=len)
uzunluk = [len(c) for c in liste]
ort_uzunluk = sum(uzunluk) / len(uzunluk)

print("Toplam yorum sayısı : ", cumle_sayisi)
print("'iyi' geçen yorum sayisi :", kac_iyi)
print("En uzun yorum : ", en_uzun)
print("En kısa yorum : ", en_kisa)
print("Ortalama uzunluk : ", ort_uzunluk)
print("Girdiğiniz Yorum Listesi :", liste )
```
